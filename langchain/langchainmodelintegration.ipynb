#### **Models Integration with OpenAI, Google Gemini and Groq**

```python
import os
from dotenv import load_dotenv
load_dotenv() ##Loads the environment variables from a .env file into the system's environment variables

os.environ["ABACUS_API_KEY"]=os.getenv("ABACUS_API_KEY")
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model="gpt-4o",
    api_key=os.environ["ABACUS_API_KEY"],
    base_url="https://routellm.abacus.ai/v1",
)
```

```python
llm.invoke("What is the date today?")
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model(
    "gpt-4o",
    api_key=os.environ["ABACUS_API_KEY"],
    base_url="https://routellm.abacus.ai/v1"
)
model
```

```python
model.invoke("What is the weather in South Dakota?")
```

#### **Google Gemini Model Integration**

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")

model = init_chat_model(
    "google_genai:gemini-3.5-flash-lite",
    api_key=os.environ["GOOGLE_API_KEY"],
)

response = model.invoke("Why do parrots talk?")
response.content
```

```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
response = model.invoke("Why do parrots talk?")
response
```

#### **Groq Model Integration**

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model(model="groq:openai/gpt-oss-120b")
response = model.invoke("Why it is raining in Boston?")
response
```

```python
import os
from langchain_groq import ChatGroq

model = ChatGroq(model="openai/gpt-oss-120b") ##Do not need to provide the provider.
response = model.invoke("Why it is raining in Boston?")
response
```

#### **Streaming Response**

```python
for chunk in model.stream("Write a poem about the beauty of nature."):
        print(chunk.text, end="|", flush=True)

```

```python
model.invoke("Write a poem about the beauty of nature.").text
```

#### **Batch Reponses**

```python
responses = model.batch(["Write a poem about the beauty of nature.", 
                         "Write a poem about the mother.",
                         "Write a rap song about the father."])

for response in responses:
    print(response.text)
```

```python
model.batch(["Write a poem about the beauty of nature.", 
                         "Write a poem about the mother.",
                         "Write a rap song about the father.",
                         "Write a rap song about the country.",
                         "Write a response to the professor.",
                         "Write a response to the minister.",
                         "Write a response to the president."],
    config={
        'max_concurrency': 5,  # Limit to 5 parallel calls (taken 5 calls at a time)
    }
)
```
